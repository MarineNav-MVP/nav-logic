# ⚓ Training YOLO11n for Day Shapes Detection
This notebook performs training on the synthetic marine day shapes dataset.

## Steps:
1. Install Ultralytics.
2. Connect Google Drive and extract the dataset.
3. Train YOLO11n model.
4. Export weights to Google Drive.

```python
# 1. Environment Setup
!pip install ultralytics
```

```python
# 2. Data Retrieval
from google.colab import drive
from pathlib import Path

drive.mount("/content/drive")

zip_path = "/content/drive/MyDrive/marine_day_shapes.zip"
if Path(zip_path).exists():
    !cp {zip_path} .
    !unzip -q marine_day_shapes.zip
    print("✅ Dataset extracted successfully.")
else:
    print("❌ Dataset zip not found on Drive!")
```

```python
# 3. Training
from ultralytics import YOLO

# Load a pretrained YOLO11n model
model = YOLO("yolo11n.pt") 

results = model.train(
    data="marine_day_shapes/data.yaml",
    epochs=50,
    imgsz=1024,
    batch=16,
    name="yolo11_day_shapes_run"
)
```

```python
# 4. Export Results
import shutil
source = "runs/detect/yolo11_day_shapes_run"
destination = "/content/drive/MyDrive/yolo11_day_shapes_results"

shutil.copytree(source, destination, dirs_exist_ok=True)
print(f"✅ Weights and logs saved to: {destination}")
```
